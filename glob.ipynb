```python
import glob as gb
```

```python
for file in gb.glob(pathname= r"D:\**" , recursive= True):
    print(file)
```

```python
for file in gb.glob(pathname= r"D:\*" , recursive= True, include_hidden= False):
    print(file)
```

```python
for file in gb.glob(pathname= r"D:\..\**" , recursive= True, include_hidden= False):
    print(file)
```
