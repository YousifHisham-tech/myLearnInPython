```python
import torch
import cv2
import os

from torchvision.models.detection import fasterrcnn_resnet50_fpn
from torchvision.models.detection.faster_rcnn import FastRCNNPredictor
from torchvision.transforms.functional import to_tensor


# ============================================================
# SETTINGS
# ============================================================

MODEL_PATH = r"Table_detection_model.pth"

INPUT_FOLDER = r"Data before preprocessing"

OUTPUT_FOLDER = r"detected_tables"

CONFIDENCE_THRESHOLD = 0.5


# ============================================================
# DEVICE
# ============================================================

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", device)

if device.type == "cuda":
    print("GPU:", torch.cuda.get_device_name(0))


# ============================================================
# CREATE FASTER R-CNN
# ============================================================

model = fasterrcnn_resnet50_fpn(
    weights=None
)

# 0 = background
# 1 = table

num_classes = 2

in_features = (
    model.roi_heads
    .box_predictor
    .cls_score
    .in_features
)

model.roi_heads.box_predictor = FastRCNNPredictor(
    in_features,
    num_classes
)


# ============================================================
# LOAD TRAINED CHECKPOINT
# ============================================================

checkpoint = torch.load(
    MODEL_PATH,
    map_location=device,
    weights_only=False
)

model.load_state_dict(
    checkpoint["model_state_dict"]
)

model.to(device)

model.eval()

print("Model loaded successfully!")


# ============================================================
# DETECT TABLES IN ONE IMAGE
# ============================================================

def detect_tables(image_path):

    image = cv2.imread(image_path)

    if image is None:
        print(f"Could not read: {image_path}")
        return None, []

    # BGR → RGB
    image_rgb = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2RGB
    )

    # Convert to tensor
    image_tensor = to_tensor(image_rgb)

    # Move to GPU
    image_tensor = image_tensor.to(device)

    # Prediction
    with torch.no_grad():

        prediction = model(
            [image_tensor]
        )[0]

    tables = []

    for box, label, score in zip(
        prediction["boxes"],
        prediction["labels"],
        prediction["scores"]
    ):

        score = score.item()
        label = label.item()

        if (
            label == 1
            and score >= CONFIDENCE_THRESHOLD
        ):

            x1, y1, x2, y2 = (
                box.cpu().tolist()
            )

            tables.append({
                "box": [
                    int(x1),
                    int(y1),
                    int(x2),
                    int(y2)
                ],
                "confidence": score
            })

    return image, tables


# ============================================================
# PROCESS ALL IMAGES
# ============================================================

os.makedirs(
    OUTPUT_FOLDER,
    exist_ok=True
)


image_extensions = (
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".tif",
    ".tiff"
)


image_files = [
    file
    for file in os.listdir(INPUT_FOLDER)
    if file.lower().endswith(image_extensions)
]


print()
print("Number of images:", len(image_files))
print()


# ============================================================
# LOOP THROUGH IMAGES
# ============================================================

for image_number, filename in enumerate(
    image_files,
    start=1
):

    image_path = os.path.join(
        INPUT_FOLDER,
        filename
    )

    print(
        f"[{image_number}/{len(image_files)}] "
        f"Processing: {filename}"
    )


    # Detect tables

    image, tables = detect_tables(
        image_path
    )


    if image is None:
        continue


    print(
        f"Tables detected: {len(tables)}"
    )


    # ========================================================
    # Create folder for this page
    # ========================================================

    page_name = os.path.splitext(
        filename
    )[0]

    page_output_folder = os.path.join(
        OUTPUT_FOLDER,
        page_name
    )

    os.makedirs(
        page_output_folder,
        exist_ok=True
    )


    # ========================================================
    # Crop every detected table
    # ========================================================

    height, width = image.shape[:2]


    for table_number, table in enumerate(
        tables,
        start=1
    ):

        x1, y1, x2, y2 = table["box"]


        # Make sure coordinates are valid

        x1 = max(0, min(x1, width))
        x2 = max(0, min(x2, width))

        y1 = max(0, min(y1, height))
        y2 = max(0, min(y2, height))


        # Crop table

        table_crop = image[
            y1:y2,
            x1:x2
        ]


        # Save table

        output_path = os.path.join(
            page_output_folder,
            f"table_{table_number}.png"
        )


        cv2.imwrite(
            output_path,
            table_crop
        )


        print(
            f"  Table {table_number}: "
            f"{table['confidence']:.3f}"
        )


    print()
```

```python
import cv2
import pytesseract
import pandas as pd
import os



# ============================================================
# 2. SETTINGS
# ============================================================

TABLE_IMAGE = r"detected_tables\page_11\table_3.png"

OUTPUT_CSV = r"table_2.csv"

LANGUAGE = "eng"


# ============================================================
# 3. READ TABLE IMAGE
# ============================================================

image = cv2.imread(TABLE_IMAGE)

if image is None:
    raise ValueError(
        f"Could not read image: {TABLE_IMAGE}"
    )


# ============================================================
# 4. CONVERT BGR → GRAYSCALE
# ============================================================

gray = cv2.cvtColor(
    image,
    cv2.COLOR_BGR2GRAY
)


# ============================================================
# 5. CONVERT GRAYSCALE → BINARY
# ============================================================

threshold_value, binary = cv2.threshold(
    gray,
    0,
    255,
    cv2.THRESH_BINARY + cv2.THRESH_OTSU
)

print(
    "Otsu threshold:",
    threshold_value
)


# ============================================================
# 6. SAVE BINARY IMAGE (OPTIONAL)
# ============================================================

cv2.imwrite(
    "binary_table.png",
    binary
)


# ============================================================
# 7. OCR USING TESSERACT
# ============================================================

data = pytesseract.image_to_data(
    binary,
    lang=LANGUAGE,
    config="--psm 6",
    output_type=pytesseract.Output.DATAFRAME
)


# ============================================================
# 8. REMOVE EMPTY OCR RESULTS
# ============================================================

data = data.dropna(
    subset=["text"]
)

data["text"] = data["text"].astype(str)

data = data[
    data["text"].str.strip() != ""
]


# ============================================================
# 9. SORT OCR RESULTS
# ============================================================

data = data.sort_values(
    ["top", "left"]
)


# ============================================================
# 10. GROUP WORDS INTO ROWS
# ============================================================

rows = []

current_row = []

current_y = None

Y_TOLERANCE = 15


for _, word in data.iterrows():

    x = int(word["left"])
    y = int(word["top"])

    text = word["text"].strip()


    # First word

    if current_y is None:

        current_y = y

        current_row = [
            (x, text)
        ]

        continue


    # Same row

    if abs(y - current_y) <= Y_TOLERANCE:

        current_row.append(
            (x, text)
        )


    # New row

    else:

        rows.append(
            sorted(
                current_row,
                key=lambda item: item[0]
            )
        )

        current_row = [
            (x, text)
        ]

        current_y = y


# Add the last row

if current_row:

    rows.append(
        sorted(
            current_row,
            key=lambda item: item[0]
        )
    )


# ============================================================
# 11. CONVERT ROWS TO TEXT
# ============================================================

table_rows = []

for row in rows:

    texts = [
        text
        for x, text in row
    ]

    table_rows.append(texts)


# ============================================================
# 12. PRINT OCR RESULT
# ============================================================

print("\nOCR result:\n")

for i, row in enumerate(table_rows):

    print(
        f"Row {i + 1}:",
        row
    )


# ============================================================
# 13. HANDLE EMPTY RESULT
# ============================================================

if len(table_rows) == 0:

    print(
        "No text was detected."
    )

else:

    # ========================================================
    # 14. FIND MAXIMUM NUMBER OF COLUMNS
    # ========================================================

    max_columns = max(
        len(row)
        for row in table_rows
    )


    # ========================================================
    # 15. MAKE ALL ROWS SAME LENGTH
    # ========================================================

    normalized_rows = []

    for row in table_rows:

        normalized_row = row + [
            ""
        ] * (
            max_columns - len(row)
        )

        normalized_rows.append(
            normalized_row
        )


    # ========================================================
    # 16. CREATE DATAFRAME
    # ========================================================

    df = pd.DataFrame(
        normalized_rows
    )


    # ========================================================
    # 17. SAVE CSV
    # ========================================================

    df.to_csv(
        OUTPUT_CSV,
        index=False,
        header=False,
        encoding="utf-8-sig"
    )


    print(
        f"\nCSV created successfully:"
    )

    print(
        OUTPUT_CSV
    )
```

```python
import cv2
import numpy as np
import pytesseract
import pandas as pd
import os


# ============================================================
# SETTINGS
# ============================================================

image_path = r"D:\Programming\Projects\Learning python librairies\OCR\cropped_tables\table_2.png"

output_folder = "Outline_Tables"

csv_folder = "CSV_Tables"

os.makedirs(
    output_folder,
    exist_ok=True
)

os.makedirs(
    csv_folder,
    exist_ok=True
)


# ============================================================
# 1. READ IMAGE
# ============================================================

img = cv2.imread(image_path)

if img is None:
    raise ValueError(
        f"Could not read image: {image_path}"
    )


# ============================================================
# 2. GRAYSCALE
# ============================================================

gray = cv2.cvtColor(
    img,
    cv2.COLOR_BGR2GRAY
)


# ============================================================
# 3. BINARY
#
# White = text / table lines
# Black = background
# ============================================================

_, binary = cv2.threshold(
    gray,
    0,
    255,
    cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU
)


# ============================================================
# 4. DETECT HORIZONTAL LINES
# ============================================================

horizontal_kernel_size = max(
    30,
    img.shape[1] // 30
)

horizontal_kernel = cv2.getStructuringElement(
    cv2.MORPH_RECT,
    (
        horizontal_kernel_size,
        1
    )
)

horizontal_lines = cv2.morphologyEx(
    binary,
    cv2.MORPH_OPEN,
    horizontal_kernel
)


# ============================================================
# 5. DETECT VERTICAL LINES
# ============================================================

vertical_kernel_size = max(
    30,
    img.shape[0] // 30
)

vertical_kernel = cv2.getStructuringElement(
    cv2.MORPH_RECT,
    (
        1,
        vertical_kernel_size
    )
)

vertical_lines = cv2.morphologyEx(
    binary,
    cv2.MORPH_OPEN,
    vertical_kernel
)


# ============================================================
# 6. COMBINE LINES
# ============================================================

table_lines = cv2.bitwise_or(
    horizontal_lines,
    vertical_lines
)


# ============================================================
# 7. CONNECT LINES BELONGING TO SAME TABLE
# ============================================================

connect_kernel = cv2.getStructuringElement(
    cv2.MORPH_RECT,
    (
        20,
        20
    )
)

table_mask = cv2.morphologyEx(
    table_lines,
    cv2.MORPH_CLOSE,
    connect_kernel
)


# ============================================================
# 8. FIND TABLE CONTOURS
# ============================================================

contours, _ = cv2.findContours(
    table_mask,
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_SIMPLE
)


# ============================================================
# 9. GET CANDIDATE TABLES
# ============================================================

tables = []

image_area = (
    img.shape[0] *
    img.shape[1]
)


for contour in contours:

    x, y, w, h = cv2.boundingRect(
        contour
    )

    area = w * h

    if (
        w >= 100
        and
        h >= 100
        and
        area >= 0.003 * image_area
    ):

        tables.append(
            (x, y, w, h)
        )


# ============================================================
# 10. SORT TABLES
# ============================================================

tables = sorted(
    tables,
    key=lambda box: (
        box[1],
        box[0]
    )
)


print(
    f"Detected {len(tables)} tables"
)


# ============================================================
# FUNCTION:
# GROUP NEARBY LINE POSITIONS
# ============================================================

def group_positions(
    positions,
    tolerance=5
):

    if len(positions) == 0:
        return []

    groups = []

    current_group = [
        positions[0]
    ]

    for position in positions[1:]:

        if (
            position -
            current_group[-1]
            <= tolerance
        ):

            current_group.append(
                position
            )

        else:

            groups.append(
                current_group
            )

            current_group = [
                position
            ]

    groups.append(
        current_group
    )

    return [
        int(np.mean(group))
        for group in groups
    ]


# ============================================================
# FUNCTION:
# DETECT CELLS INSIDE ONE TABLE
# ============================================================

def extract_table_cells(
    table_crop
):

    # --------------------------------------------------------
    # Grayscale
    # --------------------------------------------------------

    gray_table = cv2.cvtColor(
        table_crop,
        cv2.COLOR_BGR2GRAY
    )


    # --------------------------------------------------------
    # Binary
    # --------------------------------------------------------

    _, binary_table = cv2.threshold(
        gray_table,
        0,
        255,
        cv2.THRESH_BINARY_INV +
        cv2.THRESH_OTSU
    )


    height, width = (
        binary_table.shape
    )


    # --------------------------------------------------------
    # Horizontal lines
    # --------------------------------------------------------

    horizontal_kernel_size = max(
        20,
        width // 30
    )

    horizontal_kernel = (
        cv2.getStructuringElement(
            cv2.MORPH_RECT,
            (
                horizontal_kernel_size,
                1
            )
        )
    )

    horizontal_lines = (
        cv2.morphologyEx(
            binary_table,
            cv2.MORPH_OPEN,
            horizontal_kernel
        )
    )


    # --------------------------------------------------------
    # Vertical lines
    # --------------------------------------------------------

    vertical_kernel_size = max(
        20,
        height // 30
    )

    vertical_kernel = (
        cv2.getStructuringElement(
            cv2.MORPH_RECT,
            (
                1,
                vertical_kernel_size
            )
        )
    )

    vertical_lines = (
        cv2.morphologyEx(
            binary_table,
            cv2.MORPH_OPEN,
            vertical_kernel
        )
    )


    # --------------------------------------------------------
    # Find horizontal line coordinates
    # --------------------------------------------------------

    horizontal_projection = np.sum(
        horizontal_lines > 0,
        axis=1
    )


    horizontal_positions = (
        np.where(
            horizontal_projection
            > width * 0.20
        )[0]
    )


    horizontal_positions = (
        group_positions(
            horizontal_positions
        )
    )


    # --------------------------------------------------------
    # Find vertical line coordinates
    # --------------------------------------------------------

    vertical_projection = np.sum(
        vertical_lines > 0,
        axis=0
    )


    vertical_positions = (
        np.where(
            vertical_projection
            > height * 0.20
        )[0]
    )


    vertical_positions = (
        group_positions(
            vertical_positions
        )
    )


    return (
        gray_table,
        horizontal_positions,
        vertical_positions
    )


# ============================================================
# PROCESS EACH TABLE
# ============================================================

for table_number, (
    x,
    y,
    w,
    h
) in enumerate(
    tables,
    start=1
):

    print()
    print(
        f"Processing table {table_number}"
    )


    # ========================================================
    # CROP ORIGINAL IMAGE
    # ========================================================

    padding = 15

    x1 = max(
        x - padding,
        0
    )

    y1 = max(
        y - padding,
        0
    )

    x2 = min(
        x + w + padding,
        img.shape[1]
    )

    y2 = min(
        y + h + padding,
        img.shape[0]
    )


    table_crop = img[
        y1:y2,
        x1:x2
    ]


    # ========================================================
    # SAVE TABLE
    # ========================================================

    table_path = os.path.join(
        output_folder,
        f"table_{table_number}.png"
    )

    cv2.imwrite(
        table_path,
        table_crop
    )


    # ========================================================
    # FIND CELLS
    # ========================================================

    (
        gray_table,
        horizontal_positions,
        vertical_positions
    ) = extract_table_cells(
        table_crop
    )


    print(
        "Horizontal lines:",
        len(horizontal_positions)
    )

    print(
        "Vertical lines:",
        len(vertical_positions)
    )


    # ========================================================
    # CHECK CELLS
    # ========================================================

    if (
        len(horizontal_positions) < 2
        or
        len(vertical_positions) < 2
    ):

        print(
            "Could not detect enough "
            "lines to extract cells."
        )

        continue


    # ========================================================
    # OCR EACH CELL
    # ========================================================

    table_data = []


    for row in range(
        len(horizontal_positions) - 1
    ):

        row_data = []


        y1_cell = horizontal_positions[
            row
        ]

        y2_cell = horizontal_positions[
            row + 1
        ]


        for col in range(
            len(vertical_positions) - 1
        ):

            x1_cell = vertical_positions[
                col
            ]

            x2_cell = vertical_positions[
                col + 1
            ]


            # ------------------------------------------------
            # Remove table borders
            # ------------------------------------------------

            padding = 5


            crop_x1 = max(
                x1_cell + padding,
                0
            )

            crop_x2 = min(
                x2_cell - padding,
                gray_table.shape[1]
            )

            crop_y1 = max(
                y1_cell + padding,
                0
            )

            crop_y2 = min(
                y2_cell - padding,
                gray_table.shape[0]
            )


            # ------------------------------------------------
            # Extract cell
            # ------------------------------------------------

            cell = gray_table[
                crop_y1:crop_y2,
                crop_x1:crop_x2
            ]


            # ------------------------------------------------
            # Binary cell
            # ------------------------------------------------

            _, cell_binary = cv2.threshold(
                cell,
                0,
                255,
                cv2.THRESH_BINARY +
                cv2.THRESH_OTSU
            )


            # ------------------------------------------------
            # OCR
            # ------------------------------------------------

            text = (
                pytesseract.image_to_string(
                    cell_binary,
                    lang="eng",
                    config="--psm 6"
                )
            )


            text = (
                text
                .strip()
                .replace("\n", " ")
            )


            row_data.append(
                text
            )


        table_data.append(
            row_data
        )


    # ========================================================
    # CREATE DATAFRAME
    # ========================================================

    df = pd.DataFrame(
        table_data
    )


    # ========================================================
    # SAVE CSV
    # ========================================================

    csv_path = os.path.join(
        csv_folder,
        f"table_{table_number}.csv"
    )


    df.to_csv(
        csv_path,
        index=False,
        header=False,
        encoding="utf-8-sig"
    )


    print(
        f"CSV saved: {csv_path}"
    )


    # ========================================================
    # PRINT DATA
    # ========================================================

    print()
    print(df)
```

```python
import cv2
import numpy as np
import pytesseract
import pandas as pd
import os

# ============================================================
# SETTINGS
# ============================================================

IMAGE_PATH = r"D:\Programming\Projects\Learning python librairies\OCR\cropped_tables\table_3.png"

TABLE_OUTPUT_FOLDER = "Detected_Tables"

CSV_OUTPUT_FOLDER = "CSV_Tables"

DEBUG_OUTPUT_FOLDER = "Debug_Cells"

os.makedirs(TABLE_OUTPUT_FOLDER, exist_ok=True)
os.makedirs(CSV_OUTPUT_FOLDER, exist_ok=True)
os.makedirs(DEBUG_OUTPUT_FOLDER, exist_ok=True)


# ============================================================
# 1. READ IMAGE
# ============================================================

img = cv2.imread(IMAGE_PATH)

if img is None:
    raise ValueError(
        f"Could not read image: {IMAGE_PATH}"
    )


# ============================================================
# 2. GRAYSCALE
# ============================================================

gray = cv2.cvtColor(
    img,
    cv2.COLOR_BGR2GRAY
)


# ============================================================
# 3. BINARY
#
# White = text / table lines
# Black = background
# ============================================================

_, binary = cv2.threshold(
    gray,
    0,
    255,
    cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU
)


# ============================================================
# 4. HORIZONTAL LINES
# ============================================================

horizontal_kernel_size = max(
    30,
    img.shape[1] // 30
)

horizontal_kernel = cv2.getStructuringElement(
    cv2.MORPH_RECT,
    (
        horizontal_kernel_size,1
    )
)

horizontal_lines = cv2.morphologyEx(
    binary,
    cv2.MORPH_OPEN,
    horizontal_kernel
)


# ============================================================
# 5. VERTICAL LINES
# ============================================================

vertical_kernel_size = max(
    30,
    img.shape[0] // 30
)

vertical_kernel = cv2.getStructuringElement(
    cv2.MORPH_RECT,
    (
        1,
        vertical_kernel_size
    )
)

vertical_lines = cv2.morphologyEx(
    binary,
    cv2.MORPH_OPEN,
    vertical_kernel
)


# ============================================================
# 6. COMBINE LINES
# ============================================================

table_lines = cv2.bitwise_or(
    horizontal_lines,
    vertical_lines
)


# ============================================================
# 7. CONNECT LINES
# ============================================================

table_mask = cv2.bitwise_or(
    horizontal_lines,
    vertical_lines
)


# ============================================================
# 8. FIND TABLE CONTOURS
# ============================================================

contours, _ = cv2.findContours(
    table_mask,
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_SIMPLE
)


# ============================================================
# 9. GET TABLES
# ============================================================

tables = []

image_area = (
    img.shape[0] *
    img.shape[1]
)


for contour in contours:

    x, y, w, h = cv2.boundingRect(
        contour
    )

    area = w * h

    if (
        w >= 100
        and
        h >= 100
        and
        area >= 0.003 * image_area
    ):

        tables.append(
            (x, y, w, h)
        )


# ============================================================
# 10. SORT TABLES
# ============================================================

tables = sorted(
    tables,
    key=lambda box: (
        box[1],
        box[0]
    )
)


print(
    f"Detected {len(tables)} tables"
)


# ============================================================
# FUNCTION:
# OCR ONE CELL
# ============================================================

def ocr_cell(cell):

    # ---------------------------------------------
    # Convert to grayscale
    # ---------------------------------------------

    if len(cell.shape) == 3:

        cell_gray = cv2.cvtColor(
            cell,
            cv2.COLOR_BGR2GRAY
        )

    else:

        cell_gray = cell


    # ---------------------------------------------
    # Binary
    # ---------------------------------------------

    _, cell_binary = cv2.threshold(
        cell_gray,
        0,
        255,
        cv2.THRESH_BINARY + cv2.THRESH_OTSU
    )


    # ---------------------------------------------
    # OCR
    # ---------------------------------------------

    text = pytesseract.image_to_string(
        cell_binary,
        lang="eng",
        config="--psm 4"
    )


    # ---------------------------------------------
    # Clean text
    # ---------------------------------------------

    text = text.strip()

    text = " ".join(
        text.split()
    )


    return text


# ============================================================
# FUNCTION:
# DETECT RECTANGULAR CELLS
# ============================================================

def detect_cells(table_image):

    # ---------------------------------------------
    # Grayscale
    # ---------------------------------------------

    gray_table = cv2.cvtColor(
        table_image,
        cv2.COLOR_BGR2GRAY
    )


    # ---------------------------------------------
    # Binary
    #
    # White = lines
    # Black = background
    # ---------------------------------------------

    _, binary_table = cv2.threshold(
        gray_table,
        0,
        255,
        cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU
    )


    # ---------------------------------------------
    # Detect horizontal lines
    # ---------------------------------------------

    height, width = binary_table.shape

    horizontal_kernel_size = max(
        30,
        width // 30
    )

    horizontal_kernel = cv2.getStructuringElement(
        cv2.MORPH_RECT,
        (
            horizontal_kernel_size,
            1
        )
    )

    horizontal_lines = cv2.morphologyEx(
        binary_table,
        cv2.MORPH_OPEN,
        horizontal_kernel
    )


    # ---------------------------------------------
    # Detect vertical lines
    # ---------------------------------------------

    vertical_kernel_size = max(
        30,
        height // 30
    )

    vertical_kernel = cv2.getStructuringElement(
        cv2.MORPH_RECT,
        (
            1,
            vertical_kernel_size
        )
    )

    vertical_lines = cv2.morphologyEx(
        binary_table,
        cv2.MORPH_OPEN,
        vertical_kernel
    )


    # ---------------------------------------------
    # Combine table lines
    # ---------------------------------------------

    lines = cv2.bitwise_or(
        horizontal_lines,
        vertical_lines
    )


    # ---------------------------------------------
    # Find contours of rectangles
    # ---------------------------------------------

    contours, _ = cv2.findContours(
        lines,
        cv2.RETR_LIST,
        cv2.CHAIN_APPROX_SIMPLE
    )


    cells = []

    table_area = (
        width *
        height
    )


    # ---------------------------------------------
    # Analyze contours
    # ---------------------------------------------

    for contour in contours:

        x, y, w, h = cv2.boundingRect(
            contour
        )


        # Area

        area = w * h


        # -----------------------------------------
        # Basic size filtering
        # -----------------------------------------

        if w < 40:
            continue

        if h < 20:
            continue

        if area < 0.001 * table_area:
            continue


        # -----------------------------------------
        # Approximate contour
        # -----------------------------------------

        perimeter = cv2.arcLength(
            contour,
            True
        )

        epsilon = (
            0.02 *
            perimeter
        )

        approx = cv2.approxPolyDP(
            contour,
            epsilon,
            True
        )


        # -----------------------------------------
        # Check rectangle
        # -----------------------------------------

        if len(approx) != 4:
            continue


        # -----------------------------------------
        # Check that contour is convex
        # -----------------------------------------

        if not cv2.isContourConvex(
            approx
        ):
            continue


        # -----------------------------------------
        # Aspect ratio
        # -----------------------------------------

        aspect_ratio = w / float(h)


        # Ignore extremely thin shapes

        if aspect_ratio > 30:
            continue

        if aspect_ratio < 0.03:
            continue


        cells.append(
            (
                x,
                y,
                w,
                h
            )
        )


    # ========================================================
    # REMOVE DUPLICATE / NESTED RECTANGLES
    # ========================================================

    cells = sorted(
        cells,
        key=lambda box: (
            box[1],
            box[0]
        )
    )


    filtered_cells = []


    for cell in cells:

        x, y, w, h = cell

        x2 = x + w
        y2 = y + h


        keep = True


        for existing in filtered_cells:

            ex, ey, ew, eh = existing

            ex2 = ex + ew
            ey2 = ey + eh


            # Intersection

            ix1 = max(
                x,
                ex
            )

            iy1 = max(
                y,
                ey
            )

            ix2 = min(
                x2,
                ex2
            )

            iy2 = min(
                y2,
                ey2
            )


            if (
                ix2 > ix1
                and
                iy2 > iy1
            ):

                intersection = (
                    (ix2 - ix1) *
                    (iy2 - iy1)
                )

                current_area = (
                    w *
                    h
                )

                existing_area = (
                    ew *
                    eh
                )


                # If one rectangle almost
                # contains the other

                if (
                    intersection
                    /
                    min(
                        current_area,
                        existing_area
                    )
                    > 0.8
                ):

                    keep = False

                    break


        if keep:

            filtered_cells.append(
                cell
            )


    return filtered_cells


# ============================================================
# PROCESS EVERY TABLE
# ============================================================

for table_number, (
    x,
    y,
    w,
    h
) in enumerate(
    tables,
    start=1
):

    print()
    print(
        "=" * 60
    )

    print(
        f"TABLE {table_number}"
    )


    # ========================================================
    # CROP ORIGINAL TABLE
    # ========================================================

    padding = 15

    x1 = max(
        x - padding,
        0
    )

    y1 = max(
        y - padding,
        0
    )

    x2 = min(
        x + w + padding,
        img.shape[1]
    )

    y2 = min(
        y + h + padding,
        img.shape[0]
    )


    table_crop = img[
        y1:y2,
        x1:x2
    ]


    # ========================================================
    # SAVE TABLE
    # ========================================================

    table_path = os.path.join(
        TABLE_OUTPUT_FOLDER,
        f"table_{table_number}.png"
    )

    cv2.imwrite(
        table_path,
        table_crop
    )


    # ========================================================
    # DETECT RECTANGULAR CELLS
    # ========================================================

    cells = detect_cells(
        table_crop
    )


    print(
        f"Detected cells: {len(cells)}"
    )


    # ========================================================
    # DEBUG IMAGE
    # ========================================================

    debug = table_crop.copy()


    for cell_number, (
        cx,
        cy,
        cw,
        ch
    ) in enumerate(
        cells,
        start=1
    ):

        cv2.rectangle(
            debug,
            (cx, cy),
            (
                cx + cw,
                cy + ch
            ),
            (0, 0, 255),
            2
        )

        cv2.putText(
            debug,
            str(cell_number),
            (
                cx + 5,
                cy + 20
            ),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.6,
            (255, 0, 0),
            2
        )


    debug_path = os.path.join(
        DEBUG_OUTPUT_FOLDER,
        f"table_{table_number}_cells.png"
    )

    cv2.imwrite(
        debug_path,
        debug
    )


    # ========================================================
    # SORT CELLS INTO ROWS
    # ========================================================

    # First sort approximately by Y

    cells = sorted(
        cells,
        key=lambda box: (
            box[1],
            box[0]
        )
    )


    # ========================================================
    # GROUP CELLS INTO ROWS
    # ========================================================

    rows = []

    row_y_tolerance = 15


    for cell in cells:

        cx, cy, cw, ch = cell

        center_y = (
            cy +
            ch / 2
        )


        placed = False


        for row in rows:

            row_center_y = np.mean(
                [
                    r[1] + r[3] / 2
                    for r in row
                ]
            )


            if abs(
                center_y -
                row_center_y
            ) <= row_y_tolerance:

                row.append(
                    cell
                )

                placed = True

                break


        if not placed:

            rows.append(
                [cell]
            )


    # ========================================================
    # SORT EACH ROW LEFT → RIGHT
    # ========================================================

    for row in rows:

        row.sort(
            key=lambda box: box[0]
        )


    # Sort rows top → bottom

    rows.sort(
        key=lambda row: np.mean(
            [
                cell[1]
                for cell in row
            ]
        )
    )


    # ========================================================
    # OCR CELLS
    # ========================================================

    table_data = []


    for row_number, row in enumerate(
        rows
    ):

        row_data = []


        for col_number, cell in enumerate(
            row
        ):

            cx, cy, cw, ch = cell


            # --------------------------------------------
            # Remove rectangle border
            # --------------------------------------------

            cell_padding = 5


            crop_x1 = max(
                cx + cell_padding,
                0
            )

            crop_y1 = max(
                cy + cell_padding,
                0
            )

            crop_x2 = min(
                cx + cw - cell_padding,
                table_crop.shape[1]
            )

            crop_y2 = min(
                cy + ch - cell_padding,
                table_crop.shape[0]
            )


            # --------------------------------------------
            # Crop cell
            # --------------------------------------------

            cell_image = table_crop[
                crop_y1:crop_y2,
                crop_x1:crop_x2
            ]


            # --------------------------------------------
            # OCR
            # --------------------------------------------

            text = ocr_cell(
                cell_image
            )


            print(
                f"Row {row_number + 1}, "
                f"Column {col_number + 1}: "
                f"{text}"
            )


            row_data.append(
                text
            )


        table_data.append(
            row_data
        )


    # ========================================================
    # CREATE DATAFRAME
    # ========================================================

    if len(table_data) == 0:

        print(
            "No cells found."
        )

        continue


    max_columns = max(
        len(row)
        for row in table_data
    )


    # Make all rows same length

    for row in table_data:

        while len(row) < max_columns:

            row.append("")


    df = pd.DataFrame(
        table_data
    )


    # ========================================================
    # SAVE CSV
    # ========================================================

    csv_path = os.path.join(
        CSV_OUTPUT_FOLDER,
        f"table_{table_number}.csv"
    )


    df.to_csv(
        csv_path,
        index=False,
        header=False,
        encoding="utf-8-sig"
    )


    print()
    print(
        f"CSV saved: {csv_path}"
    )


    print()
    print(df)
```

```python
import cv2
import numpy as np
import pytesseract
import pandas as pd
import os


# ==========================================================
# 1. Group nearby positions
# ==========================================================

def group_positions(positions, min_distance=5):

    if len(positions) == 0:
        return []

    groups = []
    current_group = [positions[0]]

    for p in positions[1:]:

        if p - current_group[-1] <= min_distance:
            current_group.append(p)

        else:
            groups.append(current_group)
            current_group = [p]

    groups.append(current_group)

    # Get center of every group
    centers = [
        int(np.mean(group))
        for group in groups
    ]

    return centers


# ==========================================================
# 2. OCR one cell
# ==========================================================

def ocr_cell(cell):

    # ------------------------------------------
    # Make sure cell is grayscale
    # ------------------------------------------

    if len(cell.shape) == 3:
        cell_gray = cv2.cvtColor(
            cell,
            cv2.COLOR_BGR2GRAY
        )
    else:
        cell_gray = cell.copy()

    # ------------------------------------------
    # Threshold
    # ------------------------------------------

    _, cell_binary = cv2.threshold(
        cell_gray,
        0,
        255,
        cv2.THRESH_BINARY + cv2.THRESH_OTSU
    )

    # ------------------------------------------
    # OCR
    # ------------------------------------------

    text = pytesseract.image_to_string(
        cell_binary,
        lang="eng",
        config="--psm 6"
    )

    # ------------------------------------------
    # Clean text
    # ------------------------------------------

    text = text.strip()

    return text


# ==========================================================
# 3. Extract table cells
# ==========================================================

def extract_csv_from_image(
    input_path_image,
    output_path_csv,
    output_cells_dir=None
):

    # ======================================================
    # READ IMAGE
    # ======================================================

    img = cv2.imread(
        input_path_image,
        cv2.IMREAD_GRAYSCALE
    )

    if img is None:
        raise ValueError(
            f"Could not read image: {input_path_image}"
        )

    # ======================================================
    # BINARY
    # ======================================================

    _, binary = cv2.threshold(
        img,
        0,
        255,
        cv2.THRESH_BINARY + cv2.THRESH_OTSU
    )

    height, width = binary.shape

    # ======================================================
    # HORIZONTAL KERNEL
    # ======================================================

    horizontal_kernel_size = max(
        30,
        width // 30
    )

    horizontal_kernel = cv2.getStructuringElement(
        cv2.MORPH_RECT,
        (
            horizontal_kernel_size,
            1
        )
    )

    # ======================================================
    # VERTICAL KERNEL
    # ======================================================

    vertical_kernel_size = max(
        30,
        height // 30
    )

    vertical_kernel = cv2.getStructuringElement(
        cv2.MORPH_RECT,
        (
            1,
            vertical_kernel_size
        )
    )

    # ======================================================
    # DETECT HORIZONTAL LINES
    # ======================================================

    horizontal_lines = cv2.morphologyEx(
        binary,
        cv2.MORPH_OPEN,
        horizontal_kernel
    )

    # ======================================================
    # DETECT VERTICAL LINES
    # ======================================================

    vertical_lines = cv2.morphologyEx(
        binary,
        cv2.MORPH_OPEN,
        vertical_kernel
    )

    # ======================================================
    # COMBINE LINES
    # ======================================================

    table_lines = cv2.bitwise_or(
        horizontal_lines,
        vertical_lines
    )

    # ======================================================
    # FIND X POSITIONS OF VERTICAL LINES
    # ======================================================

    vertical_projection = np.sum(
        vertical_lines > 0,
        axis=0
    )

    vertical_positions = np.where(
        vertical_projection > 0
    )[0]

    vertical_x = group_positions(
        vertical_positions,
        min_distance=5
    )

    # ======================================================
    # FIND Y POSITIONS OF HORIZONTAL LINES
    # ======================================================

    horizontal_projection = np.sum(
        horizontal_lines > 0,
        axis=1
    )

    horizontal_positions = np.where(
        horizontal_projection > 0
    )[0]

    horizontal_y = group_positions(
        horizontal_positions,
        min_distance=5
    )

    # ======================================================
    # PRINT DETECTED LINES
    # ======================================================

    print("\nVertical lines:")
    print(vertical_x)

    print("\nHorizontal lines:")
    print(horizontal_y)

    print(
        "\nNumber of columns:",
        len(vertical_x) - 1
    )

    print(
        "Number of rows:",
        len(horizontal_y) - 1
    )

    # ======================================================
    # CHECK
    # ======================================================

    if len(vertical_x) < 2:
        raise ValueError(
            "Not enough vertical lines detected."
        )

    if len(horizontal_y) < 2:
        raise ValueError(
            "Not enough horizontal lines detected."
        )

    # ======================================================
    # CREATE OUTPUT DIRECTORY FOR CELLS
    # ======================================================

    if output_cells_dir is not None:

        os.makedirs(
            output_cells_dir,
            exist_ok=True
        )

    # ======================================================
    # 2D TABLE
    # ======================================================

    table_data = []

    # ======================================================
    # LOOP THROUGH ROWS
    # ======================================================

    for row in range(
        len(horizontal_y) - 1
    ):

        row_data = []

        # --------------------------------------------------
        # LOOP THROUGH COLUMNS
        # --------------------------------------------------

        for col in range(
            len(vertical_x) - 1
        ):

            # ==============================================
            # CELL POSITION
            # ==============================================

            x1 = vertical_x[col]
            x2 = vertical_x[col + 1]

            y1 = horizontal_y[row]
            y2 = horizontal_y[row + 1]

            print(
                f"\nCell [{row}][{col}]"
            )

            print(
                f"x1={x1}, y1={y1}, "
                f"x2={x2}, y2={y2}"
            )

            # ==============================================
            # PADDING
            # ==============================================

            # Remove table borders from OCR region
            padding = 5

            crop_x1 = x1 + padding
            crop_x2 = x2 - padding

            crop_y1 = y1 + padding
            crop_y2 = y2 - padding

            # ==============================================
            # CHECK CROP
            # ==============================================

            if (
                crop_x1 >= crop_x2
                or crop_y1 >= crop_y2
            ):

                text = ""

                row_data.append(text)

                continue

            # ==============================================
            # CROP CELL
            # ==============================================

            cell = binary[
                crop_y1:crop_y2,
                crop_x1:crop_x2
            ]

            # ==============================================
            # SAVE CELL
            # ==============================================

            if output_cells_dir is not None:

                cell_path = os.path.join(
                    output_cells_dir,
                    f"cell_r{row}_c{col}.png"
                )

                cv2.imwrite(
                    cell_path,
                    cell
                )

            # ==============================================
            # OCR
            # ==============================================

            text = ocr_cell(cell)

            # ==============================================
            # STORE TEXT
            # ==============================================

            row_data.append(text)

            print(
                f"Text: {text!r}"
            )

        # ==================================================
        # FINISHED ONE ROW
        # ==================================================

        table_data.append(row_data)

    # ======================================================
    # PRINT 2D ARRAY
    # ======================================================

    print("\n==============================")
    print("2D TABLE DATA")
    print("==============================")

    for row in table_data:
        print(row)

    # ======================================================
    # SAVE CSV
    # ======================================================

    df = pd.DataFrame(
        table_data
    )

    df.to_csv(
        output_path_csv,
        index=False,
        header=False,
        encoding="utf-8-sig"
    )

    print(
        f"\nCSV saved to: {output_path_csv}"
    )

    return table_data
extract_csv_from_image(r"D:\Programming\Projects\Learning python librairies\OCR\binary_table.png", 'test')
```

```python
import os
import cv2
import numpy as np
import pandas as pd
import pytesseract


# ==========================================================
# 1. IMAGE PREPROCESSING
# ==========================================================

def preprocess_image(image):
    """
    Convert image to grayscale and binary format.

    Output:
        gray   -> original grayscale image
        binary -> white text/lines on black background
    """

    gray = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2GRAY
    )

    _, binary = cv2.threshold(
        gray,
        0,
        255,
        cv2.THRESH_BINARY + cv2.THRESH_OTSU
    )

    binary = cv2.bitwise_not(binary)

    return gray, binary


# ==========================================================
# 2. DETECT TABLE LINES
# ==========================================================

def detect_table_lines(binary):
    """
    Detect horizontal and vertical table borders.
    """

    # ------------------------------------------
    # Horizontal lines
    # ------------------------------------------

    horizontal_kernel = cv2.getStructuringElement(
        cv2.MORPH_RECT,
        (50, 1)
    )

    horizontal_lines = cv2.morphologyEx(
        binary,
        cv2.MORPH_OPEN,
        horizontal_kernel
    )

    # ------------------------------------------
    # Vertical lines
    # ------------------------------------------

    vertical_kernel = cv2.getStructuringElement(
        cv2.MORPH_RECT,
        (1, 50)
    )

    vertical_lines = cv2.morphologyEx(
        binary,
        cv2.MORPH_OPEN,
        vertical_kernel
    )

    # ------------------------------------------
    # Combine
    # ------------------------------------------

    table_lines = cv2.bitwise_or(
        horizontal_lines,
        vertical_lines
    )

    return (
        horizontal_lines,
        vertical_lines,
        table_lines
    )


# ==========================================================
# 3. GROUP LINE POSITIONS
# ==========================================================

def group_positions(
    positions,
    distance=5
):
    """
    Merge nearby line positions into one position.
    """

    if len(positions) == 0:
        return []

    groups = []

    start = positions[0]
    previous = positions[0]

    for position in positions[1:]:

        if position - previous > distance:

            groups.append(
                (start + previous) // 2
            )

            start = position

        previous = position

    groups.append(
        (start + previous) // 2
    )

    return groups


# ==========================================================
# 4. DETECT CELL BOUNDARIES
# ==========================================================

def detect_cell_boundaries(
    horizontal_lines,
    vertical_lines,
    distance=5
):
    """
    Find the row and column boundaries of the table.
    """

    # ------------------------------------------
    # Horizontal projection
    # ------------------------------------------

    horizontal_projection = np.sum(
        horizontal_lines > 0,
        axis=1
    )

    horizontal_positions = np.where(
        horizontal_projection > 0
    )[0]

    # ------------------------------------------
    # Vertical projection
    # ------------------------------------------

    vertical_projection = np.sum(
        vertical_lines > 0,
        axis=0
    )

    vertical_positions = np.where(
        vertical_projection > 0
    )[0]

    # ------------------------------------------
    # Group nearby positions
    # ------------------------------------------

    horizontal_positions = group_positions(
        horizontal_positions,
        distance
    )

    vertical_positions = group_positions(
        vertical_positions,
        distance
    )

    return (
        horizontal_positions,
        vertical_positions
    )


# ==========================================================
# 5. CROP ONE CELL
# ==========================================================

def crop_cell(
    gray,
    x1,
    y1,
    x2,
    y2,
    margin=5
):
    """
    Crop one cell while removing its borders.
    """

    x1 = max(
        x1 + margin,
        0
    )

    x2 = min(
        x2 - margin,
        gray.shape[1]
    )

    y1 = max(
        y1 + margin,
        0
    )

    y2 = min(
        y2 - margin,
        gray.shape[0]
    )

    if (
        x2 <= x1
        or
        y2 <= y1
    ):
        return None

    return gray[
        y1:y2,
        x1:x2
    ]


# ==========================================================
# 6. OCR ONE CELL
# ==========================================================

def ocr_cell(
    cell,
    lang="eng+ara"
):
    """
    Extract text from one table cell.
    """

    if cell is None:
        return ""

    text = pytesseract.image_to_string(
        cell,
        lang=lang,
        config="--psm 6"
    )

    # Clean whitespace
    text = " ".join(
        text.split()
    )

    return text


# ==========================================================
# 7. EXTRACT TABLE CELLS
# ==========================================================

def extract_table_cells(
    gray,
    horizontal_positions,
    vertical_positions,
    margin=5,
    lang="eng+ara"
):
    """
    Extract every cell and OCR its contents.
    """

    table_data = []

    # ------------------------------------------
    # Rows
    # ------------------------------------------

    for row in range(
        len(horizontal_positions) - 1
    ):

        row_data = []

        y1 = horizontal_positions[row]
        y2 = horizontal_positions[row + 1]

        # --------------------------------------
        # Columns
        # --------------------------------------

        for col in range(
            len(vertical_positions) - 1
        ):

            x1 = vertical_positions[col]
            x2 = vertical_positions[col + 1]

            # ----------------------------------
            # Crop cell
            # ----------------------------------

            cell = crop_cell(
                gray,
                x1,
                y1,
                x2,
                y2,
                margin
            )

            # ----------------------------------
            # OCR
            # ----------------------------------

            text = ocr_cell(
                cell,
                lang
            )

            row_data.append(
                text
            )

        table_data.append(
            row_data
        )

    return table_data


# ==========================================================
# 8. CLEAN TABLE DATA
# ==========================================================

def clean_table_data(
    table_data
):
    """
    Convert extracted data to DataFrame
    and remove completely empty rows/columns.
    """

    df = pd.DataFrame(
        table_data
    )

    # Replace empty strings
    # with NaN temporarily
    df = df.replace(
        "",
        np.nan
    )

    # Remove empty rows
    df = df.dropna(
        how="all"
    )

    # Remove empty columns
    df = df.dropna(
        axis=1,
        how="all"
    )

    # Convert NaN back to ""
    df = df.fillna("")

    return df


# ==========================================================
# 9. SAVE CSV
# ==========================================================

def save_table_csv(
    dataframe,
    output_path
):
    """
    Save table DataFrame as CSV.
    """

    os.makedirs(
        os.path.dirname(output_path),
        exist_ok=True
    )

    dataframe.to_csv(
        output_path,
        index=False,
        header=False,
        encoding="utf-8-sig"
    )


# ==========================================================
# 10. PROCESS ONE CROPPED TABLE
# ==========================================================

def process_table(
    image_path,
    output_csv,
    margin=5,
    lang="eng+ara"
):
    """
    Complete pipeline for one cropped table.
    """

    # ------------------------------------------
    # Read image
    # ------------------------------------------

    image = cv2.imread(
        image_path
    )

    if image is None:
        print(
            f"Could not read: {image_path}"
        )
        return

    # ------------------------------------------
    # Preprocessing
    # ------------------------------------------

    gray, binary = preprocess_image(
        image
    )

    # ------------------------------------------
    # Detect lines
    # ------------------------------------------

    (
        horizontal_lines,
        vertical_lines,
        table_lines
    ) = detect_table_lines(
        binary
    )

    # ------------------------------------------
    # Detect cell boundaries
    # ------------------------------------------

    (
        horizontal_positions,
        vertical_positions
    ) = detect_cell_boundaries(
        horizontal_lines,
        vertical_lines
    )

    # ------------------------------------------
    # Validate
    # ------------------------------------------

    if (
        len(horizontal_positions) < 2
        or
        len(vertical_positions) < 2
    ):

        print(
            f"Could not detect cells: {image_path}"
        )

        return

    # ------------------------------------------
    # OCR cells
    # ------------------------------------------

    table_data = extract_table_cells(
        gray,
        horizontal_positions,
        vertical_positions,
        margin,
        lang
    )

    # ------------------------------------------
    # DataFrame
    # ------------------------------------------

    df = clean_table_data(
        table_data
    )

    # ------------------------------------------
    # Save
    # ------------------------------------------

    save_table_csv(
        df,
        output_csv
    )

    print(
        f"Saved: {output_csv}"
    )

    return df


# ==========================================================
# 11. PROCESS ENTIRE FOLDER
# ==========================================================

def process_table_folder(
    input_folder,
    output_folder
):
    """
    Process all cropped table images.
    """

    os.makedirs(
        output_folder,
        exist_ok=True
    )

    extensions = (
        ".png",
        ".jpg",
        ".jpeg",
        ".bmp",
        ".tif",
        ".tiff"
    )

    files = sorted(
        f
        for f in os.listdir(input_folder)
        if f.lower().endswith(
            extensions
        )
    )

    print(
        f"Found {len(files)} table images."
    )

    for filename in files:

        image_path = os.path.join(
            input_folder,
            filename
        )

        name = os.path.splitext(
            filename
        )[0]

        output_csv = os.path.join(
            output_folder,
            f"{name}.csv"
        )

        print(
            f"\nProcessing: {filename}"
        )

        process_table(
            image_path,
            output_csv
        )

    print(
        "\nAll tables processed."
    )
process_table_folder(
    input_folder=r"D:\Programming\Projects\Learning python librairies\OCR\cropped_tables",
    output_folder=r"D:\Programming\Projects\Learning python librairies\OCR\CSV_Tables"
)
```

```python
def preprocess_image(image):
    """
    Accept:
        - BGR image
        - RGB image
        - Grayscale image
        - Binary image

    Return:
        gray   -> grayscale image
        binary -> white text/lines on black background
    """

    # ======================================================
    # 1. Check image
    # ======================================================

    if image is None:
        raise ValueError("Image is None")

    # ======================================================
    # 2. Determine image type
    # ======================================================

    # ------------------------------------------------------
    # Grayscale / Binary
    # ------------------------------------------------------

    if len(image.shape) == 2:

        gray = image.copy()

    # ------------------------------------------------------
    # RGB / BGR
    # ------------------------------------------------------

    elif len(image.shape) == 3:

        channels = image.shape[2]

        if channels == 3:

            # IMPORTANT:
            # OpenCV normally reads images as BGR.
            #
            # For grayscale conversion, RGB vs BGR makes
            # very little difference for normal document
            # images, but we assume OpenCV/BGR here.

            gray = cv2.cvtColor(
                image,
                cv2.COLOR_BGR2GRAY
            )

        elif channels == 4:

            # BGR + Alpha

            gray = cv2.cvtColor(
                image,
                cv2.COLOR_BGRA2GRAY
            )

        else:

            raise ValueError(
                f"Unsupported number of channels: {channels}"
            )

    else:

        raise ValueError(
            f"Unsupported image shape: {image.shape}"
        )

    # ======================================================
    # 3. Make sure grayscale is uint8
    # ======================================================

    if gray.dtype != np.uint8:

        gray = cv2.normalize(
            gray,
            None,
            0,
            255,
            cv2.NORM_MINMAX
        ).astype(np.uint8)

    # ======================================================
    # 4. Determine whether image is already binary
    # ======================================================

    unique_values = np.unique(gray)

    is_binary = np.all(
        np.isin(
            unique_values,
            [0, 255]
        )
    )

    # ======================================================
    # 5. Binary conversion
    # ======================================================

    if is_binary:

        binary = gray.copy()

    else:

        _, binary = cv2.threshold(
            gray,
            0,
            255,
            cv2.THRESH_BINARY + cv2.THRESH_OTSU
        )

    # ======================================================
    # 6. Make:
    #
    # White = text / table lines
    # Black = background
    # ======================================================

    # If background is white, invert it.
    #
    # We determine this from the image borders.
    # Document background is normally found around the
    # corners.

    border_pixels = np.concatenate([
        binary[0, :],
        binary[-1, :],
        binary[:, 0],
        binary[:, -1]
    ])

    background_value = np.median(
        border_pixels
    )

    if background_value > 127:

        binary = cv2.bitwise_not(
            binary
        )

    return gray, binary
```
