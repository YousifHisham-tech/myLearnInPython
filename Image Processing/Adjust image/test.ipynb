```python
import cv2
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("jonah.png",0)
rock = cv2.imread("rock.png",0)
```

```python
plt.imshow(rock, cmap='grey')
```

```python
clache = cv2.createCLAHE(clipLimit = 3.0, tileGridSize = (4,4))
greyrock = clache.apply(rock)
plt.imshow(greyrock, cmap='grey')
```

```python
clache = cv2.createCLAHE(clipLimit = 3.0, tileGridSize = (4,4))
img2 = clache.apply(img)
```

```python
plt.imshow(img, cmap = 'grey')
```

```python
plt.imshow(img2, cmap='grey')
```

```python
plt.hist(img2)
```
