```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import io, img_as_float
from PIL import Image
import cv2
import os
```

```python
img = io.imread('image.jpg') # IO make the image into an numpy array
print(img.shape)
plt.imshow(img)
plt.show()
```

```python
img = Image.open('image.jpg')
print("1)",img.format)
print("2)",img.size)
print("3)",img.mode)
print("4)",img.info)
print("5)",img.getbands())
print ("6)",img.getbbox())
print ("7)",img.filename)
print ("8)",img.getdata())
print ("9)",img.getpixel((img.size[0] / 2, img.size[1] / 2)))
print ("10)",img.getcolors())
print ("11)",img.getextrema())
img.show()
```

```python
from PIL import Image
from PIL.ExifTags import TAGS

img = Image.open("image.jpg")
exif_data = img.getexif()

for tag_id in exif_data:
  tag = TAGS.get(tag_id, tag_id)
  data = exif_data.get(tag_id)
  print(tag, data)
```
