```python
from skimage import io
from skimage.filters.rank import entropy
from skimage.morphology import disk
from skimage.filters import threshold_mean, threshold_otsu, threshold_local, threshold_yen, threshold_triangle, threshold_niblack, threshold_sauvola
import matplotlib.pyplot as plt
from scipy import misc
import numpy as np
```

```python
img = misc.face(gray=True)
entr_img = entropy(img, disk(6))
plt.imshow(entr_img, cmap='gray')
plt.axis('off')
plt.colorbar()
```

```python
img = misc.face(gray=True)
entr_img = entropy(img, disk(5))
plt.imshow(255-entr_img, cmap='gray')
plt.axis('off')
plt.colorbar()
```

```python
plt.imshow(img, cmap='gray')
plt.axis('off')
plt.colorbar()
plt.show()
```

```python
# convert the image to binary using a global thresholding method
thre = threshold_mean(img)
plt.imshow(thre > img, cmap='binary')
plt.axis('off')
```

```python
# convert the image to binary using a global thresholding method
thre = threshold_otsu(img) #best thresholding method
plt.imshow(thre > img, cmap='binary')
plt.colorbar()
plt.axis('off')
```

```python
print(np.sum(thre > img))
print(np.sum(thre <= img))
print(np.sum(thre > img) + np.sum(thre <= img))
```
