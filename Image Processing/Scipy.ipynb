```python
from scipy import misc, ndimage
import matplotlib.pyplot as plt
from PIL import Image
from scipy.stats import linregress # used for plotting the histogram
import numpy as np
```

```python
img = misc.face()
plt.imshow(img)
```

```python
img = Image.fromarray(img)
img = img.resize((100, 100))
plt.imshow(img.resize((100, 100)))
plt.show()
```

```python
imh = ndimage.rotate(img, 45)
plt.imshow(imh)
```

```python
imh = ndimage.rotate(img, 45, reshape=False)
plt.imshow(imh)
```

```python
imgrotated = ndimage.rotate(img, 180)
plt.imshow(imgrotated)
```

```python
imgud = np.fliplr(img)
plt.imshow(imgud)
```

```python
imgud = np.flipud(img)
plt.imshow(imgud)
```

```python
img = img.convert("L")
```

```python
img_uniform_filter = ndimage.uniform_filter(img, size=10)
plt.imshow(img_uniform_filter, cmap="gray")
```

```python
img_uniform_filter = ndimage.gaussian_filter(img, sigma=3)
plt.imshow(img_uniform_filter, cmap="gray")
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 5, 4, 5]
result = linregress(x, y)
print(result)
print(result.slope)
print(result.intercept)
print (result.rvalue)
print (result.pvalue)
print (result.stderr)
print (result.intercept_stderr)
print ("r2:", result.rvalue**2)
```
