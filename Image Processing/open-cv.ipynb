```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
img = cv2.imread('image.jpg')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(img)
plt.axis('off')
```

```python
kernel = np.ones((13,13), np.float32) / np.square(13)
blur = cv2.filter2D(img, -1, kernel)
plt.imshow(blur)
plt.axis('off')
plt.show()
```

```python
blur = cv2.blur(img, (13,13))
plt.imshow(blur)
plt.axis('off')
plt.show()
```

```python
blur = cv2.GaussianBlur(img, (13,13), sigmaX=13, sigmaY=13) # 13 is the sigmaX, which controls the amount of blur. A higher value will result in a blurrier image.
plt.imshow(blur)
plt.axis('off')
plt.show()
```

```python
blur = cv2.bilateralFilter(img, d=13, sigmaColor=75, sigmaSpace=75)
plt.imshow(blur)
plt.axis('off')
plt.show()
```

```python
blur = cv2.medianBlur(img, 13)
plt.imshow(blur)
plt.axis('off')
plt.show()
```

```python
edge_detection = cv2.Canny(img, 50, 200) # used for edge detection. The first parameter is the input image, the second parameter is the lower threshold for the hysteresis procedure, and the third parameter is the upper threshold for the hysteresis procedure.
plt.imshow(edge_detection, cmap='gray')
plt.axis('off')
plt.show()
```

```python
cv2.imshow('Image', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import numpy as np
import cv2
img = cv2.imread('image.jpg')
hsv_img = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
cv2.namedWindow("Trackbars")
cv2.createTrackbar("Lower Hue", "Trackbars", 0, 179, lambda x: None)
cv2.createTrackbar("Lower Saturation", "Trackbars", 0, 255, lambda x: None)
cv2.createTrackbar("Lower Value", "Trackbars", 0, 255, lambda x: None)
cv2.createTrackbar("Upper Hue", "Trackbars", 179, 179, lambda x: None)
cv2.createTrackbar("Upper Saturation", "Trackbars", 255, 255, lambda x: None)
cv2.createTrackbar("Upper Value", "Trackbars", 255, 255, lambda x: None) 
while True:
    lower_hue = cv2.getTrackbarPos("Lower Hue", "Trackbars")
    lower_saturation = cv2.getTrackbarPos("Lower Saturation", "Trackbars")
    lower_value = cv2.getTrackbarPos("Lower Value", "Trackbars")
    upper_hue = cv2.getTrackbarPos("Upper Hue", "Trackbars")
    upper_saturation = cv2.getTrackbarPos("Upper Saturation", "Trackbars")
    upper_value = cv2.getTrackbarPos("Upper Value", "Trackbars")
    
    lower_bound = np.array([lower_hue, lower_saturation, lower_value])
    upper_bound = np.array([upper_hue, upper_saturation, upper_value])
    
    mask = cv2.inRange(hsv_img, lower_bound, upper_bound)
    result = cv2.bitwise_and(img, img, mask=mask)
    
    cv2.imshow("Result", result)
    
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
```
